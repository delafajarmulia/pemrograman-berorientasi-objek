## Praktikum 1 - Enkapsulasi pada Kelas Bank Account

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.__owner = owner
        self.__balance = balance
        
    def deposit(self, amount):
        """Method to deposit money into the account."""
        if amount > 0:
            self.__balance += amount
            print(f'{amount} telah ditambahkan ke akun {self.__owner}. Saldo saat ini: {self.__balance}')
        else:
            print('Jumlah deposit harus lebih besar dari 0.')
            
    def withdraw(self, amount):
        """Method untuk menarik uang dari akun."""
        if amount <= self.__balance:
            self.__balance -= amount
            print(f'{amount} telah ditarik dari akun {self.__owner}. Saldo saat ini: {self.__balance}')
        else:
            print('Saldo tidak cukup untuk melakukan penarikan.')
            
    def get_balace(self):
        """Method untuk mendapatkan saldo saat ini."""
        return self.__balance
    
# contoh penggunaan
if __name__ == "__main__":
    alice_account = BankAccount("Alice", 1000)
    
    alice_account.deposit(500)
    alice_account.withdraw(-100)
    
    alice_account.withdraw(300)
    alice_account.withdraw(2000)
    
    current_balance = alice_account.get_balace()
    print(f'Saldo terakhir di akun {alice_account._BankAccount__owner} adalah: {current_balance}')
    
    # akses attribute private secara langsung (tidak disarankan)
    # alice_account.__balance = 5000  # Ini tidak akan mengubah saldo karena __balance adalah private
    # print(alice_account.get_balace())  # Ini akan tetap menunjukkan saldo yang benar
```

## Praktikum 2 - Enkapsulasi pada Kelas Employee

```python
class Employee:
    def __init__(self, name, salary):
        self.__name = name
        self.__salary = salary
        
    def update_salary(self, increase):
        """Method untuk menaikkan gaji dengan validasi"""
        if increase > 0:
            self.__salary += increase
            print(f'Gaji {self.__name} telah dinaikkan sebesar {increase}. Gaji saat ini: {self.__salary}')
        else:
            print("Nilai kenanikkan harus lebih dari 0")
            
    def set_salary(self, new_salary):
        """Method untuk mengubah gaji dengan validasi"""
        if new_salary >= 0:
            self.__salary = new_salary
            print(f'Gaji diatur ulang menjadi {new_salary}')
        else:
            print("Gaji tidak boleh bernilai negatif")
            
    def get_salary(self):
        """Method untuk mendapatkan informasi gaji"""
        return self.__salary
    
    def get_employee_info(self):
        return f'Nama: {self.__name}, Gaji: {self.__salary}'
    
# contoh penggunaan
if __name__ == "__main__":
    employee1 = Employee("John Doe", 50000)
    
    print(employee1.get_employee_info())
    
    employee1.update_salary(5000)
    print(f'Gaji setelah kenaikan: {employee1.get_salary()}')
    
    employee1.set_salary(60000)
    print(f'Info terbaru: {employee1.get_employee_info()}')
    
    # mengakses attribute private secara langsung (tidak disarankan)
    # employee1.__salary = 100000  # Ini tidak akan mengubah gaji karena __salary adalah private
    # print(employee1.__salary) # Ini akan menghasilkan error karena __salary adalah private
```

## Praktikum 3 - Konstruktor dan Destruktor Sederhana

```python
class SimpleExample:
    def __init__(self, name):
        self.name = name  # Ini adalah attribute public
        print(f'Nama yang diberikan: {self.name}')
        
    def __del__(self):
        print(f'Instance dengan nama {self.name} telah dihapus.')
        
def main():
    print("Membuat instance SimpleExample...\n")
    
    obj = SimpleExample("Demo")
    print("program seddang berjalan...\n")
    
    # menghapus instance secara eksplisit
    del obj
    print("Instance telah dihapus, secara eksplisit.\n")
    print("Program selesai.")
    
if __name__ == "__main__":
    main()
```

## Praktikum 4 - Konstruktor dan Destruktor program FileLogger

```python
class FileLogger:
    def __init__(self, filename):
        self.filename = filename
        try:
            self.file = open(self.filename, 'a') # Membuka file dalam mode append
            print(f'File {self.filename} berhasil dibuka untuk logging.')
        except Exception as e:
            print(f'Gagal membuka file {self.filename}: {e}')
            
    def write_log(self, message):
        """Method untuk menulis log ke file."""
        self.file.write(message + '\n')
        self.file.flush()  # Memastikan data ditulis ke disk
        print(f'Log "{message}" telah ditulis ke file {self.filename}.')
        
    def __del__(self):
        if hasattr(self, 'file') and not self.file.closed:
            self.file.close()
            print(f'File {self.filename} telah ditutup.')
            
if __name__ == "__main__":
    logger = FileLogger("application.log")
    
    logger.write_log("aplikasi dimulai a")
    logger.write_log("melakukan operasi a...")
    logger.write_log("aplikasi a selesai")
    logger.write_log("aplikasi akan segera selesai")
    
    del logger
    
    # jika objek tidah dihapus secara eksplisit, desktruktor akan dipanggil saat program berakhir, 
    # tetapi ini tidak dapat diandalkan untuk menutup file dengan benar. Oleh karena itu, disarankan untuk menggunakan context manager atau memastikan bahwa objek dihapus secara eksplisit.
```

## Praktikum 5 - Properti "self" pada Kelas

```python
class Calculator:
    def __init__(self, initial_value = 0):
        self.value = initial_value
        
    def add(self, number):
        self.value += number
        print(f'Nilai setelah penambahan: {self.value}')
    
    def subtract(self, number):
        self.value -= number
        print(f'Nilai setelah pengurangan: {self.value}')
        
    def reset(self):
        self.value = 0
        print('Nilai telah direset ke 0.')
        
    def show_value(self):
        print(f'Nilai saat ini: {self.value}')
        
def main():
    calc1 = Calculator(10)
    
    calc1.add(5)
    calc1.subtract(3)
    calc1.show_value()
    
    calc2 = Calculator()
    calc2.add(20)
    calc2.show_value()
    
    
if __name__ == "__main__":
    main()
```

## Praktikum 6 - Menggunakan getter, setter, dan dekorator @property untuk mengelola akses atribut

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
        
    @property
    def name(self):
        """Getter untuk name, mengembalikan nilai dari __name"""
        return self.__name
    
    @name.setter
    def name(self, value):
        """Setter untuk name, mengatur nilai __name dengan validasi"""
        if not value:
            print("Nama tidak boleh kosong.")
        else:
            self.__name = value
            print(f'Nama telah diubah menjadi: {self.__name}')
            
    @property
    def age(self):
        """Getter untuk age, mengembalikan nilai dari __age"""
        return self.__age
    
    @age.setter
    def age(self, value):
        """Setter untuk age, mengatur nilai __age dengan validasi"""
        if value < 0:
            print("Usia tidak boleh negatif.")
        else:
            self.__age = value
            print(f'Usia telah diubah menjadi: {self.__age}')
            
def main():
    person = Person("Alice", 30)
    print(f'Nama: {person.name}, Usia: {person.age}')
    
    person.name = "Bob"
    person.age = 35
    print(f'Nama: {person.name}, Usia: {person.age}')
    
    person.age = -5  # Ini akan memicu validasi dan tidak mengubah usia
    
if __name__ == "__main__":
    main()
```

## Penugasan

```python
class Student:
    def __init__(self, name, score):
        self.__name = name
        self.__score = None
        self.__grade = None
        self.score = score  # Menggunakan setter untuk mengatur nilai score dan grade
        
    @property
    def name(self):
        return self.__name
    
    @name.setter
    def name(self, value):
        if not value:
            print("Nama tidak boleh kosong.")
        else:
            self.__name = value
            print(f'Nama telah diubah menjadi: {self.__name}')
    
    @property
    def score(self):
        return self.__score
    
    @score.setter
    def score(self, value):
        if 0 <= value <= 100:
            self.__score = value
            self.__update_grade()
        else: 
            print("Nilai harus antara 0 dan 100.")
            
    @property
    def grade(self):
        return self.__grade
    
    def __update_grade(self):
        """Method untuk memperbarui grade berdasarkan score"""
        if self.__score >= 90:
            self.__grade = 'A'
        elif self.__score >= 80:
            self.__grade = 'B'
        elif self.__score >= 70:
            self.__grade = 'C'
        elif self.__score >= 60:
            self.__grade = 'D'
        else:
            self.__grade = 'E'
            
    def show_info(self):
        print(f'Nama Mahasiswa: {self.__name}')
        print(f'Nilai: {self.__score}')
        print(f'Grade: {self.__grade}')
        
    def __del__(self):
        print(f'Instance Student dengan nama {self.__name} telah dihapus.')
        

mhs = Student("Siti", 87)
mhs.show_info()

print("\nNilai diubah...")
mhs.score = 92
mhs.show_info()

print("\nMenghapus instance...")
del mhs
```
