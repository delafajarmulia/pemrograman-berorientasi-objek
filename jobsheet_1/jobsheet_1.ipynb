## Praktikum 1: Tipe Data Dasar

```python
# 1. Number (int, float, complex)
# Integer (int)
angka_int = 10
print(f'angka_int: {angka_int}, -> {type(angka_int)}')

# float
angka_float = 3.14
print(f'angka_float: {angka_float}, -> {type(angka_float)}')

# complex
angka_complex = 3 + 4j
print(f'angka_complex: {angka_complex}, -> {type(angka_complex)}')

# 2. Boolean
is_active = True
print(f'is_active {is_active}, -> {type(is_active)}')

# 3. string
teks = "Hello, World!"
print(f'teks: {teks}, -> {type(teks)}')


# 4. list
# list adalah tipe data terurut dan dapat diubah (mutable)
daftar_angka = [1, 2, 3, 4, 5]
print(f'daftar_angka: {daftar_angka}, -> {type(daftar_angka)}')

# 5. tuple
# tuple adalah tipe data terurut dan tidak dapat diubah (immutable)
koordinat = (10, 20)
print(f'koordinat: {koordinat}, -> {type(koordinat)}')

# 6. dictionary
# dictionary adalah tipe data berpasangan (key-value) dan dapat diubah (mutable)
data_mahasiswa = {
    "nama": "John",
    "nim": "4.33.25.2.07",
    "jurusan": "Informatika"
}
print(f'data_mahasiswa: {data_mahasiswa}, -> {type(data_mahasiswa)}')

# 7. set
# set adalah tipe data yang tidak terurut, unik, dapat diubah (mutable)
himpunan_angka = {1, 2, 3, 2, 1}
print(f'himpunan_angka: {himpunan_angka}, -> {type(himpunan_angka)}')

# 8. contoh penggunaan konversi tipe data
nilai_str = "100"
print(f'\nnilai_str: {nilai_str}, -> {type(nilai_str)}')
nilai_int = int(nilai_str) # konversi string to int
print(f'nilai_int: {nilai_int}, -> {type(nilai_int)}')
```

## Praktikum 2: Variabel dan Operasi Dasar

```python
# 1. pendeklarasian variabel
nama = "Dela"
umur = 19
tinggi = 153.1
is_student = True

print(f'nama = {nama}')
print(f'umur = {umur}')
print(f'tinggi = {tinggi}')
print(f'is_student = {is_student}')

# python tidak memerlukan deklarasi tipe data variable secara eksplisit

# 2. operasi aritmetika
a = 10
b = 3

penjumlahan = a + b
pengurangan = a - b
perkalian = a * b
pembagian = a / b
pembagian_bulat = a // b
modulus = a % b
pangkat = a ** b

print(f'\nOPERASI ARITMETIKA')
print(f'penjumlahan = {penjumlahan}')
print(f'pengurangan = {pengurangan}')
print(f'perkalian = {perkalian}')
print(f'pembagian = {pembagian}')
print(f'pembagian_bulat = {pembagian_bulat}')
print(f'modulus = {modulus}')
print(f'pangkat = {pangkat}')

# 3. operasi perbandingan
# menghasilkan nilai boolean
lebih_besar = a > b
kurang_dari = a < b
sama_dengan = a == b
tidak_sama = a != b
lebih_besar_sama = a >= b
kurang_dari_sama = a <= b

print(f'\nOPERASI PERBANDINGAN')
print(f'lebih_besar = {lebih_besar}')
print(f'kurang_dari = {kurang_dari}')
print(f'sama_dengan = {sama_dengan}')
print(f'tidak_sama = {tidak_sama}')
print(f'lebih_besar_sama = {lebih_besar_sama}')
print(f'kurang_dari_sama = {kurang_dari_sama}')

# operasi logika
# and, or, not
x = True
y = False

logika_and = x and y
logika_or = x or y
logika_not_x = not x

print(f'\nOPERASI LOGIKA')
print(f'x = {x}')
print(f'y = {y}')
print(f'logika_and = {logika_and}')
print(f'logika_or = {logika_or}')
print(f'logika_not_x = {logika_not_x}')

# 5. contoh penggunaan di dalam percabangan
if a > b and b > 0:
    print("a lebih besar dari b dan b positif")
else:
    print("kondisi tidak terpenuhi dan b <= 0")
```

## Praktikum 3: Percabangan

```python
# 1. if sederhana
nilai = 85
print(f'Contoh IF Sederhana')
if nilai > 80:
    print(f'Selamat anda lulus dengan nilai tinggi: {nilai}')
    
# 2. if-else
umur = 17
print('Contoh IF-ELSE')
if umur >= 18:
    print(f'Anda sudah cukup umur untuk mendapatkan SIM')
else:
    print(f'Anda belum cukup umur untuk mendapatkan SIM')
    
# 3. if-elif-else
hari = "Rabu"
print('Contoh IF-ELIF-ELSE')
if hari == "Senin":
    print("Hari Senin - Saatnya kembali bekerja")
elif hari == "Selasa":
    print("Hari Selasa - Jadwal rapat mingguan")
elif hari == "Rabu":
    print("Hari Rabu - Saatnya belajar")
else:   
    print("Hari lainnya - Atur jadwalmu dengan baik.\n")

# 4. if bersarang (nested if)
suhu = 35
print("Contoh IF bersarang (nested if)")
if suhu > 30:
    print("Cuaca cukup panas.")
    if suhu > 40:
        print("Bahkan sangat panas! Disarankan banyak minum air.")
    else:
        print("Masih relatif normal, tapi tetap jaga kesehatan.")
else:
    print("Cuaca tidak terlalu panas.")
```
