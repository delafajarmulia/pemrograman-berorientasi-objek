```python
class Kucing:
    def __init__(self, nama, umur, ras):
        self.__nama = nama
        self.__umur = umur
        self.ras = ras

    def bersuara(self):
        print(f'{self.__nama} bersuara meow')

    def makan(self, makanan):
        print(f'kucing makan {makanan}')

    def __tidur(self):
        print("masih tidur jangan ganggu")

if __name__ == "__main__":
    my_kucing = Kucing("Ichi", 2, "Kucing kuning")

    my_kucing.bersuara()
    my_kucing.makan("wiskas")
    my_kucing.__tidur()

    print(f'kucing saya ras nya {my_kucing.ras}')
```
