## Praktikum 1 - Membuat dan Menggunakan Fungsi

```python
# fungsi build in
# fungsi print adalah contoh fungsi build in yang sudah tersedia di python
print("ini adalah contoh fungsi build in")

# fungsi len () adalah contoh fungsi build in untuk menghitunga panjang suatu objek
kata = "Pemrograman"
panjang_kata = len(kata)
print(f'panjang kata "{kata}" adalah {panjang_kata}')

# fungsi max() adalah contoh fungsi build in untuk mencari nilai maksimum dari suatu objek
angka = [10, 5, 30, 40, 25]
nilai_max = max(angka)
print(f'nilai maksimum dari {angka} adalah {nilai_max}')

# fungsi user-defined
# fungsi dengan satu parameter (menerima satu input)
def cek_kuadrat(angka):
    print(f'kuadrat dari {angka} adalah {angka ** 2}')
    return
    
# fungsi dengan beberapa parameter (menerima beberapa input)
def hitung_luas_persegi_panjang(panjang, lebar):
    return panjang * lebar

# fungsi dengan bebrapa tipe parameter (menerima beberapa input dengan tipe data yang berbeda)
def info_mahasiswa(nama, umur, ipk):
    print(f'Nama: {nama}, Umur: {umur} tahun, IPK: {ipk}')
    return

# fungsi tanpa return value (non-returning function)
def sapa_pengguna(nama):
    print(f'Halo, {nama}! Selamat datang di dunia Python tapi bukan ular.')
    
# fungsi dengan return value (returning function)
def hitung_keliling_persegi(sisi):
    return 4 * sisi

# pemanggilan fungsi
cek_kuadrat(5)

luas = hitung_luas_persegi_panjang(10, 5)
print(f'Luas persegi panjang dengan panjang 10 dan lebar 5 adalah {luas}')

info_mahasiswa("Budi", 22, 3.8)
sapa_pengguna("Andi")

keliling = hitung_keliling_persegi(7)
print(f'Keliling persegi dengan sisi 7 adalah {keliling}')
```

## Praktikum 3 - Membuat Kelas Sederhana

```python
class Buku:
    def __init__(self, judul, pengarang, tahun_terbit):
        self.judul = judul
        self.pengarang = pengarang
        self.tahun_terbit = tahun_terbit
        self.status = "Tersedia"
        
    def tampilkan_info(self):
        print(f'Judul: {self.judul}')
        print(f'Pengarang: {self.pengarang}')
        print(f'Tahun Terbit: {self.tahun_terbit}')
        print(f'Status: {self.status}')
        
    def pinjam(self):
        if self.status == "Tersedia":
            self.status = "Dipinjam"
            print(f"Buku '{self.judul}' berhasil dipinjam.")
        else:
            print(f"Buku '{self.judul}' sedang tidak tersedia untuk dipinjam.")
            
    def kembalikan(self):
        if self.status == "Dipinjam":
            self.status = "Tersedia"
            print(f"Buku '{self.judul}' berhasil dikembalikan.")
        else:
            print(f"Buku '{self.judul}' tidak sedang dipinjam.")
            
# membuat objek buku
buku1 = Buku("Python untuk Pemula", "Andi", 2020)
buku2 = Buku("Belajar Data Science", "Budi", 2021)

# menggunakan metode pada objek buku
buku1.tampilkan_info()
buku2.pinjam()

buku2.kembalikan()
buku1.pinjam()
buku1.tampilkan_info()

# Class mahasiswa untuk mreperesentasikan data mahasiswa
class Mahasiswa:
    # constructor untuk menginisialisasi atribut mahasiswa
    def __init__(self, nama, nim, umur):
        self.nama = nama
        self.nim = nim
        self.umur = umur
        
    def tampilkan_info(self):
        print(f'Nama: {self.nama}')
        print(f'NIM: {self.nim}')
        print(f'Umur: {self.umur} tahun')
        
    def ubah_umur(self, umur_baru):
        self.umur = umur_baru
        
# membuat objek (instance) mahasiswa
mahasiswa1 = Mahasiswa("Andi", "123456", 20)
mahasiswa2 = Mahasiswa("Budi", "654321", 22)

mahasiswa1.tampilkan_info()
print()
mahasiswa2.tampilkan_info()

mahasiswa1.ubah_umur(21)
print(f'\nSetelah mengubah umur, informasi mahasiswa1:')
mahasiswa1.tampilkan_info()
```

## Praktikum 4 - Membuat dan Menggunakan Method dalam Kelas

```python
class Mobil:
    def __init__(self, merk, warna, tahun, harga):
        self.merk = merk
        self.warna = warna
        self.tahun = tahun
        self.harga = harga
        
    def tampilkan_info(self):
        print(f'Mobil {self.merk} berwarna {self.warna}, tahun {self.tahun}, dengan harga {self.harga}')
        
    def diskon(self, persen_diskon):
        diskon_harga = self.harga * (persen_diskon / 100)
        harga_setelah_diskon = self.harga - diskon_harga
        print(f'Harga setelah diskon {persen_diskon}%: {harga_setelah_diskon}')
        
    def hitung_usia(self, tahun_sekarang):
        usia = tahun_sekarang - self.tahun
        return usia
    
    def perbarui_harga(self, harga_baru, tahun_baru):
        self.harga = harga_baru
        self.tahun = tahun_baru
        print(f'Harga dan tahun mobil {self.merk} telah diperbarui menjadi {self.harga} dan {self.tahun}.')
        
    
# Membuat objek mobil
mobil1 = Mobil("Toyota", "Hitam", 2015, 300000000)
mobil2 = Mobil("Honda", "Merah", 2018, 250000000)

# Menggunakan metode tanpa return value
mobil1.tampilkan_info()
mobil2.tampilkan_info()
# Menggunakan metode dengan satu parameter (diskon)
mobil1.diskon(10)
mobil2.diskon(15)
# Menggunakan metode dengan return value (hitung usia mobil)
usia_mobil1 = mobil1.hitung_usia(2025)
usia_mobil2 = mobil2.hitung_usia(2025)
print(f"Usia mobil1 pada tahun 2025: {usia_mobil1} tahun")
print(f"Usia mobil2 pada tahun 2025: {usia_mobil2} tahun")
```

## Penugasan

```python
# Modul untuk menghitung biaya parkir
def hitung_biaya(lama_jam, tarif_per_jam=2000):
    return lama_jam * tarif_per_jam

def sapa_pengendara(plat_nomor):
    print(f"Kendaraan {plat_nomor} terdaftar di sistem")
```

```python
class Motor:
    def __init__(self, plat_nomor, merk):
        self.plat_nomor = plat_nomor
        self.merk = merk
        self.status = "Parkir" # Status awal saat objek dibuat
        
    def tampilkan_status(self):
        print(f"Motor: {self.merk} [{self.plat_nomor}] | Status: {self.status}")

    def keluar_parkir(self, durasi):
        if self.status == "Parkir":
            self.status = "Sudah Keluar"
            total = hitung_biaya(durasi)
            print(f"\nMotor {self.plat_nomor} keluar.")
            print(f"Durasi: {durasi} jam. Total Bayar: Rp{total}")
        else:
            print(f"Motor {self.plat_nomor} sudah tidak ada di parkiran.")

def main():
    print("=== SISTEM PARKIR RODA DUA ===\n")
    
    # 1. Membuat objek motor (Instansiasi)
    motor1 = Motor("B 1234 ABC", "Vario")
    motor2 = Motor("D 9999 XYZ", "NMAX")
    
    # 2. Menggunakan fungsi dari modul
    sapa_pengendara(motor1.plat_nomor)
    sapa_pengendara(motor2.plat_nomor)
    
    print("\n--- Kondisi Saat Ini ---")
    motor1.tampilkan_status()
    motor2.tampilkan_status()
    
    # 3. Proses keluar parkir (Mengubah status & hitung biaya)
    print("\n--- Proses Transaksi ---")
    motor1.keluar_parkir(3)  # Motor 1 parkir 3 jam
    
    print("\n--- Status Akhir ---")
    motor1.tampilkan_status()
    motor2.tampilkan_status()

if __name__ == "__main__":
    main()
```
